```python
import yfinance
```

```python
yfinance.Ticker("AAPL").history("6mo")
```

```python
tiker = input("Digite el codigo de accion: \n ")
data = yfinance.Ticker(ticker).history("6mo")
data
```

```python
cierre = data.Close

cierre.plot()
```

## Crear los analisis solicitados

```python
valor_maximo =round(cierre.max(),2)
valor_minimo =round(cierre.min(),2)
valor_promedio =round(cierre.mean(),2)
```

```python
print(valor_minimo)
print(valor_maximo)
print(valor_promedio)
```
